Objective: The main goal of this project is to build a Machine Learning model that can classify house prices into different categories (e.g., Low, Medium, High) based on various features like area, rooms, location, etc. This helps in understanding how data science can be applied to real estate.

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
```



*   We check the first few rows (df.head()) to understand what kind of columns (features) are present.
*   We look for missing or null values so we can clean the data before feeding it to the model.







```python

df = pd.read_csv("/content/train.csv", keep_default_na=False, na_values=[""])
for c in df.columns.drop("SalePrice"):
    if df[c].dtype == object:
        as_num = pd.to_numeric(df[c].replace("NA", np.nan), errors="coerce")
        if as_num.notna().sum() + (df[c] == "NA").sum() == len(df) and as_num.notna().sum() > 0:
            df[c] = as_num          # numeric column that uses "NA" for missing
print(df.shape)
df.head()
```



*   We use matplotlib or seaborn to plot graphs (like bar charts or histograms).
*   This helps us see the distribution of different house price categories and understand which features might be most important.



```python
labels = ["Low", "Medium", "High"]
y = pd.qcut(df["SalePrice"], 3, labels=labels).astype(str)
print(df.groupby(y)["SalePrice"].agg(["min", "max", "count"]).loc[labels])
df["SalePrice"].hist(bins=50); plt.title("SalePrice distribution"); plt.show()
```



*   We separate our features ($X$) and the target variable ($y$).



```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

X = df.drop(columns=["Id", "SalePrice"])
num_cols = X.select_dtypes(include=np.number).columns.tolist()
cat_cols = [c for c in X.columns if c not in num_cols]
pre = ColumnTransformer([
    ("num", Pipeline([("imp", SimpleImputer(strategy="median")), ("sc", StandardScaler())]), num_cols),
    ("cat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")), ("oh", OneHotEncoder(handle_unknown="ignore"))]), cat_cols),
])
print(len(num_cols), "numeric,", len(cat_cols), "categorical")
```

We split the data into Training Set (80%) and Testing Set (20%) using train_test_split so the model can learn from training data and be tested on unseen data.

Scaling ensures that all features are on the same scale (mean = 0, variance = 1). This prevents features with larger numerical values from dominating the machine learning model.

```python
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
cv = StratifiedKFold(5, shuffle=True, random_state=42)
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "KNN": KNeighborsClassifier(n_neighbors=7),
    "Decision Tree": DecisionTreeClassifier(max_depth=6, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "SVM (RBF)": SVC(kernel="rbf", C=3, gamma="scale"),
}
rows, fitted = [], {}
for name, m in models.items():
    pipe = Pipeline([("pre", pre), ("clf", m)])
    cvs = cross_val_score(pipe, Xtr, ytr, cv=cv, n_jobs=-1)
    pipe.fit(Xtr, ytr); p = pipe.predict(Xte)
    rows.append({"Model": name, "CV acc": cvs.mean(), "Test acc": accuracy_score(yte, p),
                 "Test macro-F1": f1_score(yte, p, average="macro")})
    fitted[name] = pipe
pd.DataFrame(rows).sort_values("CV acc", ascending=False).round(4)
```

1. Model Prediction using Support Vector Machine (SVM - RBF Kernel)

Concept: After training various machine learning models, the best-performing model—specifically Support Vector Machine with a Radial Basis Function (RBF) kernel—is selected (fitted["SVM (RBF)"]).

Execution: The model makes predictions on unseen test data (Xte) using the .predict() method, generating predicted labels (p) to evaluate how well it generalizes to new data.

2. Classification Report

Concept: The classification_report function provides a comprehensive breakdown of the model's predictive performance across each individual house price category (labels=labels).

Key Metrics Included:

Precision: Measures out of all predicted positive cases, how many were actually correct.

Recall (Sensitivity): Measures out of all actual positive cases, how many the model managed to correctly predict.

F1-Score: The harmonic mean of Precision and Recall, providing a balanced overall measure of the model's accuracy per class.

3. Confusion Matrix and Visualization

Concept: A Confusion Matrix is a tabular layout that visualizes the performance of the classification algorithm by comparing actual labels (yte) against predicted labels (p).

Visualization (ConfusionMatrixDisplay):

The code uses ConfusionMatrixDisplay.from_predictions with a blue color map (cmap="Blues") to plot a clean heatmap.

Why it matters: It helps instantly identify where the model is performing well (diagonal values showing correct predictions) and where it is getting confused (off-diagonal values showing misclassifications between price categories).

```python
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
best = fitted["SVM (RBF)"]
p = best.predict(Xte)
print(classification_report(yte, p, labels=labels))
ConfusionMatrixDisplay.from_predictions(yte, p, labels=labels, cmap="Blues"); plt.show()
```

Model Choice (Random Forest): We used Random Forest because it is an ensemble learning method that combines multiple decision trees to improve accuracy and prevent overfitting.

Feature Importance: Random Forest allows us to extract feature importance scores (feature_importances_), helping us identify which house features (like area, number of rooms, etc.) affect the price the most.  

 Evaluation: We evaluated the model using Accuracy and a Classification Report to see how well it classifies different price brackets

```python
rf = fitted["Random Forest"]
names = [n.split("__", 1)[1] for n in rf.named_steps["pre"].get_feature_names_out()]
imp = pd.Series(rf.named_steps["clf"].feature_importances_, index=names).sort_values(ascending=False)
imp.head(12)[::-1].plot.barh(title="Top 12 features"); plt.show()
```
